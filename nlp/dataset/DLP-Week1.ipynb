```python
def count_tokens(example):
    example["num tokens"] = len(example["tokens"])
    return example

ds = ds.map(count_tokens)
```

```python
total_tokens = 0

for split in ds:
    total_tokens += sum(ds[split]["num tokens"])

total_tokens
```

```python
total_tokens_millions = total_tokens / 1000000
total_tokens_millions
```

```python
from datasets import concatenate_datasets
```

```python
ds = concatenate_datasets([
    ds["train"],
    ds["test"],
    ds["validation"]
])
```

```python
def make_text(example):
    example["text"] = " ".join(example["tokens"])
    return example

ds = ds.map(make_text)
```

```python
ds = ds.remove_columns(["ner_tags", "tokens"])
```

```python
def at_least_six_tokens(example):
    return len(example["text"].split()) >= 6

ds = ds.filter(at_least_six_tokens)
```
